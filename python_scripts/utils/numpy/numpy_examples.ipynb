### Numpy examples

```python
import numpy as np
```

#### Basics

```python
np.arange(9).reshape((3,3))
```

```python
np.arange(3, 12).reshape(3, 3)
```

#### Matrix multiplication

```python
m1 = [1,2]
m2 = [3,4]
```

```python
M1 = np.array(m1)
M2 = np.array(m2).reshape((2,1))
print(M1@M2)
```

```python
M2 = M2.reshape((1, 2))
print(np.outer(M1, M2))
```

```python
m1, m2 = [[1,2],
          [6,5],
          [4,1],
          [3,8]], [
              [4,8,9],
              [2,3,1]
          ]
```

```python
print(m1)
print(m2)
```

```python
np.dot(m1, m2)
```

```python
M1 = np.array(m1)
M2 = np.array(m2)
```

```python
print(M1)
print(M2)
```

```python
print(M1 @ M2)
print(np.dot(m1,m2))
```

#### Outer Product

```python
np.outer(M1, M2)
```

```python
M1 = np.array([[3],
               [2],
               [3]])
M2 = np.array([[2],
               [1]])
```

```python
np.dot(M1, M2) # dimensões diferentes n colunas primeira tem quer ser igual ao n linhas da segunda
```

```python
np.outer(M1, M2)
```

```python
A1 = np.array([1,2,3,4])
A2 = np.array([5,6,7,8,5,7])

A3 = np.outer(A1,A2)
print(A3)
```

#### Cross product

```python
m1 = [1, 3, 4]
m2 = [9, 4, 0]
```

```python
m1 = [[1,0],
      [0,1]]

m2 = [[1,2],
      [3,4]]
```

```python
cross1 = np.cross(m1, m2)
cross2 = np.cross(m2, m1)
print(cross1)
print(cross2)
```

#### Determinant

```python
M = np.array([[1,-2,4],
              [7,13,-5],
              [3,2,9]])
```

```python
detM = np.round(float(np.linalg.det(M)), 5)
print(detM)
```

```python
M = np.array([[1,5,6,7],
              [7,1,2,0],
              [1,7,6,2],
              [0,1,1,3]])
```

```python
float(np.round(np.linalg.det(M), 5))
```

#### Einstein Summation Convention

```python
m1 = [1, 3, 4]
m2 = [9, 4, 0]
```

```python
result =  np.einsum("n,n", m1, m2)
print(result)
```

```python
m1 = np.arange(9).reshape((3,3))
m2 = np.arange(3, 12).reshape((3,3))
```

```python
print(m1)
print(m2)
```

```python
np.einsum("mk, kn", m1, m2)
```

```python
print(m1 @ m2)
```

```python
int(np.einsum("mn, mn", m1, m2))
```
