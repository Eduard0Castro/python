### Número de condição

Calculado como  
\begin{aligned}\kappa(A) = \Vert A\Vert \cdot \Vert A^{-1}\Vert\end{aligned}

```python
M = np.array([[np.cos(np.pi/4)],
             [np.sin(np.pi/3)],
             [np.pi]])
M
```

```python
np.linalg.cond(M) # próximo de 1, matriz bem condicionada, pequenos erros na entrada geram pequenos erros na saída.
```

```python
cond_number = np.linalg.norm(M, 2)*np.linalg.norm(np.linalg.pinv(M), 2);print(cond_number)
# k(A) = ||M||*||M^-1|| pseudoinversa já que M não é quadrada.
```

```python
np.linalg.cond(np.array([[1,1],
                         [1,1.0001]])) # exemplo de matriz mal condicionada
```

```python

M = np.array([[-3,  2],
              [ 1,  5]])

print(sum(abs(M[:, 1])))
print(M.shape)
print(f"Norma -1 (numpy): {np.linalg.norm(M, 1)}")
norma_1 = max([sum(abs(M[:, i])) for i in range (M.shape[1])]);print(f"Norma -1: {norma_1}")
# Norma -1 é o maior valor da soma dos módulos dos elementos das colunas da matriz
```

```python
M = np.array([
    [ -3,   1,   0,   4,  -2],
    [  2,  -5,   1,  -1,   3],
    [  0,   2,  -8,   2,   1],
    [ -1,   0,   3,  -4,   6],
    [  4,   1,  -2,   1,  -5]
])

print(f"Norma -1 (numpy): {np.linalg.norm(M, 1)}")
norma_1 = max([sum(abs(M[:, i])) for i in range (M.shape[1])]);print(f"Norma -1: {norma_1}")
```

#### Norma -2 ou norma espectral

Definida como a raiz quadrada do maior valor singular da matriz, sigma max, o que equivale ao maior autovalor de Atransposta.A
Para este caso, como M não é quadrada, usa-se a decomposição SVD para calcular os valores singulares

\begin{aligned}\Vert A \Vert_2 = \sqrt{\lambda_{\max}(A^T A)}\end{aligned}

```python
K = np.transpose(M)@M

norma_2 = np.linalg.norm(M, 2);print(f"Norma -2(numpy): {norma_2}")
norma_2 = np.sqrt(max(np.linalg.svd(K, compute_uv=False))); print(f"Norma -2 na mão: {norma_2}")
# Se A fosse quadrada eu poderia calcular com o maior autovalor de A^T.A
```

Calculando o número de condição:

```python
print("Número de condição: ", cond_number := np.linalg.cond(M))
```

```python
norma_2_A = np.sqrt(max(np.linalg.svd(K, compute_uv=False)))
norma_2_A_inv = np.sqrt(max(np.linalg.svd(np.linalg.pinv(K), compute_uv=False)))

M_inv = np.linalg.pinv(M)
cond_number = np.sqrt(max(np.linalg.svd(K, compute_uv=False)))*np.sqrt(max(np.linalg.svd(np.dot(M_inv, np.transpose(M_inv)), compute_uv=False)))
print("Número de condição na mão: ", cond_number)
```
